### Q1. Find the smallest number in an array

**Question:** Find the smallest number in an array.

**Example 1:**
- Input: `arr = [5, 2, 9, 1, 7]`
- Output: `1`

**Example 2:**
- Input: `arr = [-3, -1, -7, 4]`
- Output: `-7`


```python
# Q1: Find the smallest number in an array
# Write your solution here

arr = [5, 2, 9, 1, 7]
arr = [-3, -1, -7, 4]
print(np.min(arr))

k= None
for i in arr:
    if k==None:
        k=i
    elif k>i:
        k=i
print(k)
```

### Q5. Count frequency of each element in an array

**Question:** Count frequency of each element in an array.

**Example 1:**
- Input: `arr = [1, 2, 2, 3, 1, 1]`
- Output: `1 -> 3, 2 -> 2, 3 -> 1`

**Example 2:**
- Input: `arr = [5, 5, 5]`
- Output: `5 -> 3`


```python
# Q5: Count frequency of each element in an array
# Write your solution here

arr = [1, 2, 2, 3, 1, 1]

dic={}
for i in arr:
    if i not in dic:
        dic[i]=1
    else:
        dic[i]+=1

for i in dic:
    print(i,"--->",dic[i],end=', ')
```

### Q6. Rearrange array in increasing-decreasing order

**Question:** Rearrange array in increasing-decreasing order.

**Example 1:**
- Input: `arr = [1, 2, 3, 4, 5, 6]`
- Output: `[1, 3, 5, 6, 4, 2]`

**Example 2:**
- Input: `arr = [1, 2, 3, 4, 5]`
- Output: `[1, 3, 5, 4, 2]`


```python
# Q6: Rearrange array in increasing-decreasing order
# Write your solution here

arr = [1, 4, 2, 3, 5, 6]

odd=sorted([i for i in arr if i%2==1 ])
even=sorted([i for i in arr if i%2==0], reverse=True)
print(odd+even)




arr_odd=[]
arr_even=[]
for i in arr:
    if i%2==1:
        arr_odd.append(i)

    else:
        arr_even.append(i)

print(sorted(arr_odd)+(sorted(arr_even)[::-1]))
```
