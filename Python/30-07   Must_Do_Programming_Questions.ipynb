```python
# Write a program in Python to check if a given number is prime or not.
# ● Example:
# ○ Input: 17
# ○ Output: True


def is_prime(num):
    if num==1 or num==0:
        return "Not Prime"
    for i in range(2,num):
        if num%i==0:
            return "Not Prime"
    return "Prime"

is_prime(23)
```
