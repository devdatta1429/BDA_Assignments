```python
# Write a program to check whether a given number is prime or not.

def prime(num):
    if num<=1:
        return 'Not Prime'
    else:
        for i in range(2,num):
            if num%i==0:
                return 'Not Prime'
        return 'Prime'


prime(29)
```

```python
# Write a program to print the Fibonacci series using an iterative method.m


def fib(num):
    a=0
    b=1
    for i in range(num):
        c=b
        b=a+b
        print(a, end=" ")
        a=c

fib(10)
```

```python
def fib(num):
    a = 0
    b = 1

    for i in range(num):
        print(a, end=" ")
        a, b = b, a + b

fib(10)
```

```python
# Write a program to find the factorial of a number using an iterative method.
def fact(num):
    for i in range(1,num+1):
        if num%i==0:
            print(i ,end=" ")

fact(256)
```

Write a program to find the greatest among three given numbers.

```python
def bigger(num1=0,num2=0,num3=0):
    if num1<num2:
        if num3<num2:
            return  'num2',num2
        else:
            return 'num3',num3
    else:
        return 'num1',num1

bigger(1,2,3)
```

Write a program to swap two numbers using a third variable.

```python
def swap(a,b):
    c=b
    b=a
    a=c
    return a,b

swap(1,2)
```

Write a program to check whether a given year is a leap year or not.

```python
def leaf_year(year):
    if year%100!=0 and year%4==0 or year%400==0:
        return 'leap'
    else:
        return 'not leap'

leaf_year(2003)
```
